```python

import dbldatagen as dg
import pyspark.sql.types as T

# Create a data specification
row_count = 1


spec = (
    dg.DataGenerator(spark, name="tests_data", rows=row_count)
    .withColumn("idws", "long", minValue=1, maxValue=1000)
    .withColumn("name", "string", template="User_\\d{3}")
    .withColumn("age", "integer", minValue=18, maxValue=80)
)

# Build the dataset
df = spec.build()
display(df)

# Guardar como una tabla de Delta Lake
df.write.format("delta").mode("overwrite").saveAsTable("mi_catalogo.mi_esquema.tabla_sinteticaz")

# Guardar como archivos Parquet en el almacenamiento
df.write.mode("overwrite").parquet("/mnt/datos/mi_data_sinteticaz")





```
